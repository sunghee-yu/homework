## 1-2. 문자열을 숫자로 변환하기

```python
houses = [] #1 빈 리스트 houses를 생성합니다.

for i in range(len(df_house)): #② 반복문을 사용하여 각 행을 순회합니다.
  row = df_house.iloc[i]
  house = {
      'ID' : int(row['ID']),
      'size' : int(row['size']),
      'year' : int(row['year']),
      'price' : float(row['price'])
   }
  houses.append(house)

print(houses[:3]) #③ 각 행의 값을 아래 자료형으로 변환하여 houses에 저장합니다.
df_houses = pd.DataFrame(houses)
```

##1-4. 함수와 예외 처리




```python
#① 가격을 크기로 나눈 값을 계산하는 아래 함수를 작성합니다.
def unit_price(price, size):
  try :
    price = float(price)
    size = float(size)

  except ValueError:
    print("값 오류: 숫자로 변환할 수 없는 문자열입니다.")
    return None

  except TypeError:
    print("자료형 오류: 숫자 또는 숫자 형태의 문자열을 입력하세요.")
    return None

  if size <= 0 :
   print("크기오류:",size)
   return None

  if price < 0 :
   print("가격오류:",price)
   return None


  return price/size

    # 코드 작성
```

```python
for house in houses: #⑧ 작성한 함수를 전체 houses에 적용하여 주택 ID와 크기당 가격을 출력합니다.
  result = unit_price(house['price'],house['size'])
  print(f"ID{house['ID']} 크기당 가격: {result}")
```

```python
print(unit_price(4, 40))   #⑨ 아래 호출을 순서대로 실행하여 반환값과 안내 문구를 확인합니다.
print(unit_price("4", "40"))
print(unit_price(4, 0))
print(unit_price(-4, 40))
print(unit_price("오류", 40))
print(unit_price(None, 40))
print(unit_price(6, 30))
```

##1-5. 목적에 맞는 자료구조 선택하기

```python

ids = [house['ID'] for house in houses]
print("① 전체 주택 ID:", ids)
# ⑤ 선택: 리스트(list)
#    이유: 원래 순서를 유지해야 하고, ID가 순서대로 여러 개 들어가는 작업이라 '순서 유지'가 판단 근거
#    vs 집합(set): 집합은 순서를 보장하지 않아서 '원래 순서대로' 출력할 수 없음


# ② year 값을 중복 없이 저장 → 집합
years = {house['year'] for house in houses}
print("② 건축 연도(중복 제거):", years)
# ⑤ 선택: 집합(set)
#    이유: 같은 연도가 여러 번 나와도 한 번만 저장됨 → '중복 허용 안 함'이 판단 근거
#    vs 리스트(list): 리스트는 중복을 허용해서 같은 연도가 여러 번 저장됨 (중복 제거를 직접 해야 함)


# ③ 주택 ID로 가격을 조회 → 딕셔너리
price_by_id = {house['ID']: house['price'] for house in houses}
print("③ ID 3의 가격:", price_by_id[3])
# ⑤ 선택: 딕셔너리(dict)
#    이유: ID(키)를 넣으면 가격(값)을 바로 찾을 수 있음 → '키를 통한 조회'가 판단 근거
#    vs 리스트(list): 리스트는 번호(위치)로만 꺼낼 수 있어서 ID 3의 가격을 찾으려면 처음부터 하나씩 비교해야 함


# ④ 최소 크기 30, 최대 가격 4를 변경 불가능한 묶음으로 → 튜플
condition = (30, 4)
print("④ 조건 (최소 크기, 최대 가격):", condition)
# ⑤ 선택: 튜플(tuple)
#    이유: 기준값은 실수로 바뀌면 안 됨. 튜플은 만든 뒤 추가·삭제·교체가 불가능 → '변경 불가능'이 판단 근거
#    vs 리스트(list): 리스트는 condition[0] = 20처럼 값을 바꿀 수 있어서 기준이 실수로 바뀔 위험이 있음
```
